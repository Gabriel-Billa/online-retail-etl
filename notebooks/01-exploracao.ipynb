## Análise da coluna Quantity

Foram identificados valores negativos na coluna `Quantity`.

A investigação indicou dois comportamentos distintos:

- Registros com `InvoiceNo` iniciado por `C` representam cancelamentos.
- Foram identificados 9.288 registros desse tipo.
- Foi encontrado um caso em que uma compra de 80.995 unidades foi seguida,
  12 minutos depois, pelo cancelamento das mesmas 80.995 unidades.

Também foram encontrados registros negativos sem o prefixo `C`.

- Esses registros possuem `UnitPrice = 0`.
- As descrições incluem termos como `damaged`, `check`, `thrown away`,
  `missing`, `stock check` e `adjustment`.
- Esses registros aparentam representar ajustes ou baixas de estoque,
  e não vendas.
- Parte desses registros possui `Description` nula e deverá receber
  tratamento específico durante a transformação.

Conclusão: valores negativos não devem ser simplesmente removidos,
pois representam diferentes eventos de negócio.

## Implementação e testes das regras de transformação

Após a etapa de exploração e definição das regras de negócio, as transformações identificadas foram implementadas individualmente no módulo `src/transform.py`.

Cada função foi criada e testada separadamente no notebook antes de ser incorporada ao fluxo completo de transformação.

### Funções implementadas

- `remove_duplicates(df)`  
  Remove registros completamente duplicados do dataset.

- `define_cancelamento(df)`  
  Identifica os registros de cancelamento e cria a coluna booleana `cancelado`.

- `define_ajuste_estoque(df)`  
  Identifica movimentações relacionadas a ajustes de estoque e cria a coluna booleana `ajuste_estoque`.

- `define_tipo_transacao(df)`  
  Utiliza as regras anteriores para classificar os registros em:
  - `VENDA`
  - `CANCELAMENTO`
  - `AJUSTE_DE_ESTOQUE`

- `calcula_valor_total(df)`  
  Cria a coluna `valor_total` através do cálculo:

  `Quantity × UnitPrice`

### Estratégia de desenvolvimento

As funções foram implementadas de forma incremental.

Após a criação de cada função, seu resultado foi testado no notebook antes de prosseguir para a próxima transformação.

Esse processo permitiu validar separadamente cada regra e verificar se os resultados obtidos estavam de acordo com o comportamento identificado durante a exploração dos dados.

Após os testes individuais, as funções foram reunidas em uma única função responsável por coordenar toda a etapa de transformação:

`transform_data(df)`

A função executa as transformações na sequência definida:

`remove_duplicates()`

↓  

`define_cancelamento()`

↓  

`define_ajuste_estoque()`

↓  

`define_tipo_transacao()`

↓  

`calcula_valor_total()`

↓  

**DataFrame transformado**

Com isso, as regras que inicialmente foram descobertas e testadas individualmente no notebook passaram a fazer parte de um processo de transformação reutilizável.

### Resultado

O notebook foi utilizado como ambiente de experimentação para:

**investigar → definir regras → implementar → testar**

Enquanto o módulo `src/transform.py` passou a concentrar as regras de transformação que poderão ser reutilizadas pela pipeline.

A validação da função completa `transform_data()` será realizada separadamente no notebook `02-validacao-transformacoes.ipynb`.
